```python
#This code for 5CV
import matplotlib.pyplot as plt

overlap = [0,25,50,75]

# Accuracy values
lr_4s = [60.8,61,61.7,61.8]
svm_4s = [71,73.5,75.4,80]
rf_4s = [88.7,91,93.7,97.1]

lr_8s = [63.5,65.7,65.6,67]
svm_8s = [67.8,71.2,75.2,78.5]
rf_8s = [89.9,91.9,94,97.7]

plt.figure()

plt.plot(overlap, lr_4s, marker='o', label='LR (4s)')
plt.plot(overlap, svm_4s, marker='o', label='SVM (4s)')
plt.plot(overlap, rf_4s, marker='o', label='RF (4s)')

plt.plot(overlap, lr_8s, marker='s', linestyle='--', label='LR (8s)')
plt.plot(overlap, svm_8s, marker='s', linestyle='--', label='SVM (8s)')
plt.plot(overlap, rf_8s, marker='s', linestyle='--', label='RF (8s)')

plt.xlabel("Overlap (%)")
plt.ylabel("Accuracy (%)")
plt.title("Effect of Window Overlap on Classification Accuracy")
plt.legend()
plt.grid(True)
plt.savefig("C:/Users/user/Desktop/5cvaccuracyoverlap.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

classifiers = ['LR','SVM','RF']
accuracy = [67,78.5,97.7]
f1 = [63.2,75.8,97.5]

x = range(len(classifiers))

plt.figure()

plt.bar(x, accuracy, width=0.4, label='Accuracy')
plt.bar([i+0.4 for i in x], f1, width=0.4, label='F1-score')

plt.xticks([i+0.2 for i in x], classifiers)
plt.xlabel("Classifier")
plt.ylabel("Score (%)")
plt.title(" Classier Performance Comparison(8 sec,75% overlap)")
plt.legend()
plt.savefig("C:/Users/user/Desktop/5cvbestcombo.png")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("C:/Users/user/Desktop/chartdata.csv")
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

for col in df.columns[1:]:
    plt.plot(df['subjects'], df[col], marker='o', label=col)

plt.xlabel("Subjects")
plt.ylabel("Accuracy (%)")
plt.title("Accuracy per Subject for Different Segment Sizes")
# Force all subject ticks
plt.xticks(df['subjects'], rotation=90)
plt.legend()
plt.grid(True)
plt.savefig("C:/Users/user/Desktop/LOOV_accuracy_plot.png")
plt.show()
```

```python
mean_acc = df.iloc[:, 1:].mean()

plt.figure(figsize=(8,5))
plt.bar(mean_acc.index, mean_acc.values)
plt.xticks(rotation=45)
plt.ylabel("Average Accuracy (%)")
plt.title("Average Accuracy for Each Segment Length")
plt.tight_layout()
plt.savefig("C:/Users/user/Desktop/LOOV_Avg_accuracy_plot.png")
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10,5))
sns.boxplot(data=df.iloc[:, 1:])
plt.xticks(rotation=45)
plt.ylabel("Accuracy (%)")
plt.title("Accuracy Distribution Across Subjects")
plt.savefig("C:/Users/user/Desktop/LOOV_accuracy_distrib_plot.png")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.set_index('subjects'),
            cmap='viridis',
            annot=False)

plt.title("Subject-wise Accuracy Heatmap")
plt.xlabel("Segment Length")
plt.ylabel("Subjects")
plt.savefig("C:/Users/user/Desktop/LOOV_heatmap.png")
plt.show()
```

```python
best = df.iloc[:, 1:].idxmax(axis=1)

plt.figure(figsize=(8,4))
best.value_counts().plot(kind='bar')
plt.title("Best Performing Segment Length Frequency")
plt.ylabel("Number of Subjects")
plt.show()
```
